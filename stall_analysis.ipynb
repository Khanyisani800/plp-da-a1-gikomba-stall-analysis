```python
# Task 1: Dataset
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}

# Task 1: Compute total, average, highest, and lowest daily sales for 14 days
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
max_sales = max(daily_sales)
min_sales = min(daily_sales)

max_index = daily_sales.index(max_sales)
min_index = daily_sales.index(min_sales)

best_day_name = days[max_index]
best_week_num = (max_index // 7) + 1

worst_day_name = days[min_index]
worst_week_num = (min_index // 7) + 1

print(f"Total 14-Day Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Highest Sale: KES {max_sales:,} on {best_day_name} (Week {best_week_num})")
print(f"Lowest Sale: KES {min_sales:,} on {worst_day_name} (Week {worst_week_num})")

# Task 2: Week-on-Week Totals and Growth
def week_totals(sales):
    """Returns a tuple of (week 1 total, week 2 total) using slicing."""
    return sum(sales[:7]), sum(sales[7:])

w1_tot, w2_tot = week_totals(daily_sales)
wow_growth = ((w2_tot - w1_tot) / w1_tot) * 100

print(f"Week 1 Total: KES {w1_tot:,}")
print(f"Week 2 Total: KES {w2_tot:,}")
print(f"Week-on-Week Growth: {wow_growth:.2f}%")

# Task 3: Day Classification
def classify_day(amount, average):
    """Classifies a daily sales amount based on average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"

classification_counts = {"Strong": 0, "Normal": 0, "Weak": 0}
for amount in daily_sales:
    category = classify_day(amount, avg_sales)
    classification_counts[category] += 1

print("Daily Sales Classifications:")
for cat, count in classification_counts.items():
    print(f"- {cat}: {count} days")

# Task 4: Weekdays vs. Weekends
weekday_sales = [amount for i, amount in enumerate(daily_sales) if days[i] not in ["Sat", "Sun"]]
weekend_sales = [amount for i, amount in enumerate(daily_sales) if days[i] in ["Sat", "Sun"]]

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)
diff_percentage = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales (Mon-Fri): KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales (Sat-Sun): KES {avg_weekend:,.2f}")
print(f"Weekend vs Weekday Difference: {diff_percentage:.2f}% higher on weekends")

# Task 5: Inventory Reorder & Gross Margins
def check_inventory_and_margins(stock_dict):
    """Returns products at or below reorder level and gross margin percentages."""
    reorder_list = []
    margins = {}
    for item, details in stock_dict.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(item)
        margins[item] = (details["price"] - details["cost"]) / details["price"]
    return reorder_list, margins

items_to_reorder, gross_margins = check_inventory_and_margins(stock)

print("Items needing reorder:", items_to_reorder)
print("Gross Margins:")
for item, margin in gross_margins.items():
    print(f"- {item}: {margin * 100:.2f}%")

# Task 6: Stall Performance Report
print("=" * 50)
print(f"{'STALL PERFORMANCE REPORT':^50}")
print("=" * 50)
print(f"Total 14-Day Revenue:      KES {total_sales:>15,}")
print(f"Daily Average Sales:       KES {avg_sales:>15,.2f}")
print(f"Week 1 Total Revenue:      KES {w1_tot:>15,}")
print(f"Week 2 Total Revenue:      KES {w2_tot:>15,}")
print(f"Week-on-Week Growth Rate:  {wow_growth:>18.2f}%")
print("-" * 50)
print(f"Average Weekday Sales:     KES {avg_weekday:>15,.2f}")
print(f"Average Weekend Sales:     KES {avg_weekend:>15,.2f}")
print("-" * 50)
print("Inventory Status & Margins:")
for item, details in stock.items():
    margin = (details["price"] - details["cost"]) / details["price"] * 100
    status = "REORDER NEEDED" if details["qty"] <= details["reorder"] else "OK"
    print(f"  • {item:<10} | Qty: {details['qty']:>2}kg | Margin: {margin:>5.1f}% | {status}")
print("=" * 50)
```
